# File Handling

File handling is the process of creating, reading, writing, updating, and managing files using a programming language.

In Python, file handling allows a program to store data permanently in files and access that data whenever needed.

Simple example:
A student application can use file handling to save student names and marks in a file and read them later.

# Text Files

A text file is a file used to store plain text data.

Python allows us to create, read, write, and modify text files.

Common text file extension:

`.txt`

### Real-World Example

A student management system can store student information in a text file.

For example:

Name: Manikanta

Course: Chemistry

Age: 21

```python
file = open("student.txt", "w")

file.write("Name: Manikanta\n")
file.write("Course: Chemistry\n")
file.write("Age: 21\n")

file.close()

print("File created successfully")
```

### Explanation

- `open()` opens or creates a file.
- `"student.txt"` is the file name.
- `"w"` means write mode.
- `write()` adds text to the file.
- `\n` moves the next text to a new line.
- `close()` closes the file.

# Reading a Text File

We can use Python to read the contents of a text file.

The `"r"` mode is used for reading.

### Real-World Example

A school application can read student information stored in a text file.

```python
file = open("student.txt", "r")

content = file.read()

print(content)

file.close()
```

### Explanation

- `"r"` opens the file in read mode.
- `read()` reads the complete content of the file.
- `print()` displays the content.
- `close()` closes the file after reading.

# Writing to a Text File

Writing means adding new content to a file.

The `"w"` mode is used for writing.

If the file already exists, `"w"` mode replaces its previous content.

### Real-World Example

An application can create a new report and save it to a text file.

```python
file = open("report.txt", "w")

file.write("Python Training Report")
file.write("\nCompleted File Handling")

file.close()

print("Report created")
```

### Explanation

- `"w"` opens the file in write mode.
- `write()` adds content.
- If `report.txt` does not exist, Python creates it.
- If it already exists, its old content is replaced.

# Appending to a Text File

Appending means adding new content to the end of an existing file without removing its previous content.

The `"a"` mode is used for appending.

### Real-World Example

A log file can continuously add new events without deleting previous events.

```python
file = open("student.txt", "a")

file.write("Status: Active\n")

file.close()

print("Content added")
```

### Explanation

- `"a"` opens the file in append mode.
- New content is added at the end.
- Existing content remains unchanged.
- `close()` closes the file.

# CSV Files

CSV stands for Comma-Separated Values.

CSV files are commonly used to store tabular data.

Example:

Name,Age,Course

Manikanta,21,Chemistry

Rahul,22,Physics

Python provides the built-in `csv` module for working with CSV files.

### Real-World Example

A company can store employee information in a CSV file.

```python
import csv

with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["Name", "Age", "Course"])
    writer.writerow(["Manikanta", 21, "Chemistry"])
    writer.writerow(["Rahul", 22, "Physics"])

print("CSV file created")
```

### Explanation

- `import csv` imports Python's CSV module.
- `open()` creates the CSV file.
- `newline=""` helps avoid unwanted blank lines.
- `csv.writer()` creates a CSV writer.
- `writerow()` writes one row at a time.
- The first row contains column names.
- The next rows contain student data.

# JSON Files

JSON stands for JavaScript Object Notation.

JSON is commonly used for storing and exchanging structured data.

Example:

{
    "name": "Manikanta",
    "age": 21
}

Python provides the built-in `json` module to work with JSON files.

### Real-World Example

Web applications commonly use JSON to exchange information between the frontend and backend.

```python
import json

student = {
    "name": "Manikanta",
    "age": 21,
    "course": "Chemistry"
}

with open("student.json", "w") as file:
    json.dump(student, file, indent=4)

print("JSON file created")
```

### Explanation

- `import json` imports the JSON module.
- `student` is a Python dictionary.
- `open()` creates the JSON file.
- `json.dump()` writes Python data into the JSON file.
- `indent=4` makes the JSON file easier to read.

# Pickle

Pickle is a Python module used to serialize and deserialize Python objects.

Serialization means converting a Python object into a format that can be stored.

Deserialization means converting the stored data back into a Python object.

### Real-World Example

A machine learning application can save a trained model using Pickle and load it later without training the model again.

```python
import pickle

student = {
    "name": "Manikanta",
    "age": 21
}

with open("student.pkl", "wb") as file:
    pickle.dump(student, file)

print("Object saved")
```

### Explanation

- `import pickle` imports the Pickle module.
- `student` is a Python dictionary.
- `"wb"` means write binary mode.
- `pickle.dump()` serializes and stores the object.
- The object is saved in `student.pkl`.

```python
# Reading a pickle file

with open("student.pkl", "rb") as file:
    student_data = pickle.load(file)

print(student_data)
```

# Context Manager — with

The `with` statement is used to work with resources such as files safely.

It automatically closes the file after the block finishes.

### Real-World Example

Think about borrowing a book from a library.

You use the book and return it when finished.

Similarly, `with` manages the file and closes it automatically after use.

```python
with open("student.txt", "r") as file:
    content = file.read()

print(content)
```

### Explanation

- `with open()` opens the file.
- The file is available inside the `with` block.
- `read()` reads the content.
- After leaving the block, Python automatically closes the file.
- We do not need to call `file.close()` manually.

# pathlib

`pathlib` is a Python module used to work with file and directory paths.

It provides an easy and object-oriented way to work with paths.

### Real-World Example

Instead of manually writing file paths, we can use `pathlib` to create and manage them.

```python
from pathlib import Path

file_path = Path("student.txt")

print(file_path.exists())
print(file_path.name)
```

### Explanation

- `Path` is imported from `pathlib`.
- `Path("student.txt")` creates a path object.
- `exists()` checks whether the file exists.
- `name` gives the file name.

# shutil

The `shutil` module provides functions for high-level file and directory operations.

It can be used to:

- Copy files
- Move files
- Copy directories
- Remove directories

### Real-World Example

If we want to create a backup of a file, we can use `shutil.copy()`.

```python
import shutil

shutil.copy("student.txt", "student_backup.txt")

print("File copied successfully")
```

### Explanation

- `import shutil` imports the `shutil` module.
- `shutil.copy()` copies one file to another location.
- `"student.txt"` is the source file.
- `"student_backup.txt"` is the destination file.

# File Validation

File validation means checking whether a file exists and whether it is suitable for an operation before using it.

We can use `pathlib` to perform basic file validation.

### Real-World Example

Before opening a file uploaded by a user, an application can check whether the file actually exists.

```python
from pathlib import Path

file_path = Path("student.txt")

if file_path.exists():
    print("File exists")
else:
    print("File does not exist")
```

### Explanation

- `Path()` creates a path object.
- `exists()` checks whether the file or directory exists.
- `if` checks the result.
- If the file exists, `"File exists"` is printed.
- Otherwise, `"File does not exist"` is printed.

# Directory Operations

A directory is a folder used to organize files.

Python allows us to create, check, and work with directories.

We can use modules such as `os` and `pathlib`.

### Real-World Example

A project may create separate folders for:

- Documents
- Images
- Reports
- Backups

```python
from pathlib import Path

folder = Path("reports")

folder.mkdir(exist_ok=True)

print("Directory created")
```

### Explanation

- `Path("reports")` creates a path representing the `reports` directory.
- `mkdir()` creates the directory.
- `exist_ok=True` prevents an error if the directory already exists.
