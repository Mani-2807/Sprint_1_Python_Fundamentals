## 2.1 if Statement

### Step 1 – Understand the Concept

The `if` statement is used to execute a block of code only when a specified condition is true.

Python evaluates the condition and, if the result is `True`, executes the indented statements inside the `if` block.

If the condition is `False`, the statements inside the block are skipped.

### Step 2 – Demonstrate the Concept

#### Real-World Analogy

Imagine a security system at a building.

If a person has a valid access card, the system allows them to enter.

In programming, the access-card check represents the condition, and allowing entry represents the code that runs when the condition is true.

```python
age = 21

if age >= 18:
    print("Eligible to enter")
```

### Step 3 – Explain the Implementation

The variable `age` stores the value `21`.

The condition `age >= 18` checks whether the age is at least 18.

Because the condition evaluates to `True`, Python executes the `print()` statement inside the `if` block.

## 2.2 if-else

### Step 1 – Understand the Concept

The `if-else` statement provides two possible execution paths.

If the `if` condition is true, the code inside the `if` block runs.

If the condition is false, Python executes the code inside the `else` block.

### Step 2 – Demonstrate the Concept

#### Real-World Analogy

Consider checking whether a student has passed an examination.

If the marks are greater than or equal to the passing mark, the student passes.

Otherwise, the student fails.

This gives the program two possible paths.

```python
marks = 65

if marks >= 40:
    print("Pass")
else:
    print("Fail")
```

### Step 3 – Explain the Implementation

The variable `marks` contains the student's marks.

The condition `marks >= 40` checks whether the student has reached the passing mark.

Since `65` is greater than `40`, the condition is true and the `if` block executes.

If the marks were below `40`, Python would execute the `else` block instead.

## 2.3 elif Statement

### Step 1 – Understand the Concept

The `elif` statement means "else if".

It is used when a program needs to check multiple conditions.

Python checks the conditions from top to bottom. Once it finds a condition that is true, it executes that block and skips the remaining conditions.

### Step 2 – Demonstrate the Concept

#### Real-World Example

A grading system may assign different grades depending on a student's marks.

For example:

- 90 or above → A
- 75 to 89 → B
- 60 to 74 → C
- Below 60 → D

The program needs to check multiple conditions, so `elif` is useful.

```python
marks = 82

if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 60:
    grade = "C"
else:
    grade = "D"

print("Grade:", grade)
```

### Step 3 – Explain the Implementation

The program checks the student's marks against several conditions.

First, it checks whether the marks are at least 90.

If that is false, it checks whether the marks are at least 75.

Since the value is 82, the second condition is true, so the grade becomes `B`.

The remaining conditions are not evaluated after a matching `elif` block is executed.

### 2.4 combininf `if`,`elif`,`else`


```python
temperature = 32

if temperature >= 35:
    message = "Very hot"
elif temperature >= 25:
    message = "Warm"
elif temperature >= 15:
    message = "Cool"
else:
    message = "Cold"

print(message)
```

## 2.5 for Loop

### Step 1 – Understand the Concept

A `for` loop is used to repeatedly execute a block of code for each item in an iterable.

Common iterables include lists, tuples, strings, sets, dictionaries, and ranges.

A `for` loop is useful when we know that we want to process each item in a collection or repeat an operation for a sequence of values.

### Step 2 – Demonstrate the Concept

#### Real-World Analogy

Imagine a teacher checking the names of students in a class.

Instead of writing the same instruction separately for every student, the teacher can go through the list one student at a time.

A `for` loop works similarly by processing each item one at a time.

```python
students = ["Anil", "Priya", "Rahul", "Sneha"]

for student in students:
    print("Student:", student)
```

### Step 3 – Explain the Implementation

The `students` variable contains a list of names.

The `for` loop takes each name from the list one at a time and temporarily stores it in the variable `student`.

The `print()` statement then displays the current student's name.

The loop automatically stops after all elements in the list have been processed.

### 2.6 `for` loop with `range()`

### Using range() with a for Loop

The `range()` function generates a sequence of numbers that can be used with a `for` loop.


```python
for number in range(1, 6):
    print(number)
```

### Explanation

`range(1, 6)` produces numbers starting from `1` and stopping before `6`.

Therefore, the values generated are `1, 2, 3, 4, 5`.

The `for` loop processes each value sequentially.

## 2.7 while Loop

### Step 1 – Understand the Concept

A `while` loop repeatedly executes a block of code as long as its condition remains true.

Unlike a `for` loop, a `while` loop is generally useful when repetition depends on a condition rather than directly iterating over a collection.

### Step 2 – Demonstrate the Concept

#### Real-World Analogy

Imagine filling a water tank.

You continue adding water while the tank is not full.

The condition "tank is not full" determines whether the process should continue.

A `while` loop works in a similar way.


```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

### Step 3 – Explain the Implementation

The variable `count` starts at 1.

The `while` loop checks whether `count <= 5`.

As long as the condition is true, the current value is printed.

The statement `count += 1` increases the value after each iteration.

When `count` becomes 6, the condition becomes false and the loop stops.

## 2.8 break Statement

### Step 1 – Understand the Concept

The `break` statement immediately terminates the nearest enclosing loop.

It is useful when we have found what we are looking for or when continuing the loop is no longer necessary.

### Step 2 – Demonstrate the Concept

Imagine searching for a particular book on a shelf.

Once you find the book, there is no reason to continue searching.

Similarly, `break` allows a loop to stop immediately when a required condition is satisfied.

```python
for number in range(1, 11):
    if number == 6:
        break

    print(number)
```

### Step 3 – Explain the Implementation

The loop starts with numbers from 1 to 10.

When `number` becomes 6, the condition `number == 6` becomes true.

The `break` statement immediately terminates the loop.

Therefore, numbers after 5 are not printed.

## 2.9 continue Statement

### Step 1 – Understand the Concept

The `continue` statement skips the remaining code in the current loop iteration and moves to the next iteration.

Unlike `break`, it does not terminate the entire loop.

### Step 2 – Demonstrate the Concept

Imagine processing a list of tasks but deciding to skip one task temporarily.

You continue with the remaining tasks instead of stopping the entire process.

`continue` works in a similar way.

```python
for number in range(1, 6):
    if number == 3:
        continue

    print(number)
```

### Step 3 – Explain the Implementation

The loop processes numbers from 1 to 5.

When the value becomes 3, the `continue` statement skips the `print()` statement for that iteration.

The loop then continues with 4 and 5.

Therefore, `3` is not printed, but the loop itself continues.

## 2.10 pass Statement

### Step 1 – Understand the Concept

The `pass` statement is a placeholder that does nothing when executed.

It is useful when Python requires a statement to be present syntactically, but we do not want to implement the functionality yet.

### Step 2 – Demonstrate the Concept

Imagine designing a system where we know that a particular feature will be implemented later.

We can create the structure now and use `pass` temporarily instead of leaving the block empty.

```python
age = 21

if age >= 18:
    pass

print("Program continues")
```

### Step 3 – Explain the Implementation

The `if` condition is true because the age is 21.

The `pass` statement executes but does nothing.

The program then continues to the next statement, which prints `"Program continues"`.

`pass` is therefore different from `break` and `continue`.

- `break` stops the loop.
- `continue` skips the current iteration.
- `pass` does nothing and allows execution to continue normally.

## 3.1 List

### Understand the Concept

A list is an ordered and changeable collection of elements.

Important characteristics of a list:

- Lists maintain the order of elements.
- Lists are mutable, meaning their elements can be changed.
- Lists can contain duplicate values.
- Lists can store different data types.
- List elements are accessed using indexes.
- Indexing starts from 0.

### Real-World Example

A list can represent a collection of student names:

["Anil", "Priya", "Rahul", "Sneha"]

We can add a new student, remove a student, or change a student's name.

### General Syntax

`list_name = [element1, element2, element3]`

```python
students = ["Anil", "Priya", "Rahul", "Sneha"]

print(students)
```

accessing a list elements:


```python
students = ["Anil", "Priya", "Rahul", "Sneha"]

print(students[0])
print(students[1])
print(students[3])
```

### Explanation

Python uses zero-based indexing.

Therefore:

- students[0] → first element
- students[1] → second element
- students[2] → third element
- students[3] → fourth element

modifying a list:

```python
students = ["Anil", "Priya", "Rahul", "Sneha"]

students[1] = "Kiran"

print(students)
```

adding elements:

```python
students = ["Anil", "Priya", "Rahul"]

students.append("Sneha")

print(students)
```

removing elements:

```python
students = ["Anil", "Priya", "Rahul", "Sneha"]

students.remove("Rahul")

print(students)
```

list length:

```python
students = ["Anil", "Priya", "Rahul", "Sneha"]

print("Number of students:", len(students))
```

### Implementation Explanation

In the examples above:

- `[]` is used to create a list.
- `students[0]` accesses the first element.
- `students[1] = "Kiran"` modifies an element.
- `append()` adds an element to the end.
- `remove()` removes a specific element.
- `len()` returns the number of elements.

The main feature demonstrated here is that lists are mutable.

## 3.2 Tuple

A tuple is an ordered collection of elements that cannot normally be changed after it is created.

Important characteristics:

- Tuples maintain order.
- Tuples are immutable.

## 3.5 List Comprehension

### Understand the Concept

List comprehension provides a short and readable way to create a new list from an existing iterable.

Instead of writing a complete loop, we can often create the same list in a single line.

### General Syntax

[expression for item in iterable]

A condition can also be added:

[expression for item in iterable if condition]

### Real-World Example

Suppose we want to calculate the squares of numbers from 1 to 5.

Using a normal loop requires multiple lines.

List comprehension allows us to express the same operation more concisely.

#### normal For loop


```python
numbers = [1, 2, 3, 4, 5]

squares = []

for number in numbers:
    squares.append(number ** 2)

print(squares)
```

#### List comprehension

```python
numbers = [1, 2, 3, 4, 5]

squares = [number ** 2 for number in numbers]

print(squares)
```

### Explanation

The following expression:

[number ** 2 for number in numbers]

means:

1. Take each `number` from `numbers`.
2. Calculate `number ** 2`.
3. Add the result to a new list.

List comprehensions improves readability.

## 3.6 Dictionary Comprehension

### Understand the Concept

Dictionary comprehension provides a concise way to create dictionaries.

### General Syntax

{key_expression: value_expression for item in iterable}

A condition can also be included.

### Real-World Example

Suppose we have a list of subjects and want to create a dictionary where each subject is associated with the number of characters in its name.

```python
subjects = ["Python", "SQL", "Machine Learning", "Statistics"]

subject_lengths = {
    subject: len(subject)
    for subject in subjects
}

print(subject_lengths)
```

### Explanation

The expression:

{
    subject: len(subject)
    for subject in subjects
}

means:

1. Take each subject from the `subjects` list.
2. Use the subject as the dictionary key.
3. Calculate its length.
4. Store the length as the corresponding value.

## 4.2 Arguments

### Understand the Concept

Arguments allow us to pass data into a function.

A function can receive one or more values and use those values while performing its task.

### Example

Instead of creating separate functions for different names, we can create one function that accepts a name as an argument.

This makes the function reusable.

```python
def greet(name):
    print("Hello", name)


greet("Manikanta")
greet("Anil")
greet("Priya")
```

### Explanation

In:

def greet(name):

`name` is called a parameter.

When we call:

greet("Manikanta")

`"Manikanta"` is the argument passed to the function.

The function can therefore work with different values without changing its code.

## 4.3 Multiple Arguments

A function can accept multiple parameters.

The values are assigned to parameters according to their position when positional arguments are used.

```python
def add_numbers(a, b):
    result = a + b
    return result


answer = add_numbers(10, 20)

print("Sum:", answer)
```

### Explanation

The function receives two values:

- `a`
- `b`

It calculates their sum and stores it in `result`.

The `return` statement sends the result back to the place where the function was called.

The returned value is stored in `answer`.

## 4.4 Default Arguments

### Understand the Concept

A default argument is a parameter that already has a default value.

If the caller does not provide a value, Python uses the default value.

### Real-World Example

Suppose a function displays a user's country.

We can set `"India"` as the default country.

If the user provides another country, that value will be used instead.

```python
def introduce(name, country="India"):
    print("Name:", name)
    print("Country:", country)


introduce("Manikanta")
introduce("John", "USA")
```

### Explanation

In:

country="India"

`India` is the default value.

When:

introduce("Manikanta")

is called, no country is provided, so Python uses `"India"`.

When:

introduce("John", "USA")

is called, `"USA"` replaces the default value.

## 4.5 Keyword Arguments

### Understand the Concept

Keyword arguments allow us to pass values using the parameter names.

This makes function calls more readable and allows us to provide arguments in a different order.

```python
def student_details(name, age, course):
    print("Name:", name)
    print("Age:", age)
    print("Course:", course)


student_details(
    course="Chemistry",
    name="Manikanta",
    age=21
)
```

### Explanation

Instead of relying on argument position, we explicitly specify:

course="Chemistry"
name="Manikanta"
age=21

Python matches each value with its parameter name.

This improves readability, especially when a function has many parameters.

## 4.6 *args

### Understand the Concept

`*args` allows a function to accept a variable number of positional arguments.

This is useful when we do not know in advance how many positional values will be passed.

Inside the function, `args` behaves like a tuple.

### Real-World Example

Suppose we want a function that can calculate the total of any number of values.

It could receive:

2 values

or

5 values

or

10 values.

`*args` allows this flexibility.

```python
def calculate_sum(*args):
    total = 0

    for number in args:
        total += number

    return total


print(calculate_sum(10, 20))
print(calculate_sum(10, 20, 30, 40))
```

### Explanation

The `*` before `args` allows Python to collect multiple positional arguments.

For example:

calculate_sum(10, 20, 30)

Inside the function:

args = (10, 20, 30)

The values can then be processed using a loop.

The name `args` is a convention. Another valid name could be used, but `*args` is the commonly used form.

## 4.7 **kwargs

### Understand the Concept

`**kwargs` allows a function to accept a variable number of keyword arguments.

Inside the function, the values are stored as a dictionary.

### Real-World Example

Suppose we want to create a function that can receive different pieces of student information.

Different students may have different numbers of details.

`**kwargs` allows us to accept those details flexibly.

```python
def display_details(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)


display_details(
    name="Manikanta",
    age=21,
    course="Chemistry"
)
```

### Explanation

The `**` before `kwargs` collects keyword arguments into a dictionary.

For example:

display_details(name="Manikanta", age=21)

is collected approximately as:

{
    "name": "Manikanta",
    "age": 21
}

The function then uses `.items()` to access each key and value.

Like `args`, `kwargs` is a naming convention. The important part is the `**`.

## 4.10 Recursive Functions

### Understand the Concept

A recursive function is a function that calls itself.

A recursive function needs a stopping condition called the base case.

Without a base case, the function may continue calling itself indefinitely.

### Real-World Example

Consider a countdown:

5 → 4 → 3 → 2 → 1 → Stop

A recursive function can perform this type of repeated self-call.

### Important Parts

A recursive function normally contains:

1. Base case
2. Recursive case

```python
def countdown(number):
    if number == 0:
        print("Done")
        return

    print(number)
    countdown(number - 1)


countdown(5)
```

### Explanation

The function first checks:

if number == 0:

This is the base case.

When the value reaches 0, the function stops.

Otherwise:

countdown(number - 1)

calls the same function again with a smaller value.

The sequence is:

countdown(5)
→ countdown(4)
→ countdown(3)
→ countdown(2)
→ countdown(1)
→ countdown(0)

At 0, the base case stops the recursion.

# 5.3 yield

### Definition

```python
def decorator(func):
    def wrapper(a, b):
        print("Adding numbers...")
        func(a, b)
    return wrapper

@decorator
def add(a, b):
    print(a + b)

add(10, 20)
```

## 6.1 map()

The `map()` function applies a function to every item in an iterable.

It is useful when we want to perform the same operation on every element of a collection.

Syntax:

map(function, iterable)

### Example

Suppose we have a list of numbers:

10, 20, 30, 40

If we want to double every number, we can use `map()`.

10 → 20 ,
20 → 40 ,
30 → 60 ,
40 → 80.

```python
numbers = [10, 20, 30, 40]

def double(number):
    return number * 2

result = map(double, numbers)

print(list(result))
```

## 6.2 filter()

The `filter()` function selects elements from an iterable based on a condition.

Only the elements for which the function returns `True` are included.

Syntax:

filter(function, iterable)

### Example

Suppose we have:

10, 15, 20, 25, 30

If we want only even numbers:

10 → Keep

15 → Remove

20 → Keep

25 → Remove

30 → Keep

```python
numbers = [10, 15, 20, 25, 30]

def is_even(number):
    return number % 2 == 0

result = filter(is_even, numbers)

print(list(result))
```

### Explanation

`filter()` checks every element against a condition.

If the condition returns `True`, the element is included.

If the condition returns `False`, the element is excluded.
