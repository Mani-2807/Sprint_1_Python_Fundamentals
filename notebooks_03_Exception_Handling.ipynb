# Exception Handling

### definition
Exception handling is a way of handling errors that occur while a Python program is running, so the program can respond to the error instead of stopping suddenly.

In simple words,Exception handling is like having a backup plan when something goes wrong, so the entire process doesn't stop unexpectedly.

-`try` → code that may cause an error

-`except` → handles the error

-`else` → runs if there is no error

-`finally` → runs at the end, whether there is an error or not

### Real-World Analogy

Imagine you are using an ATM:

**1.** You insert your card and enter your PIN → `try`

**2.** If the PIN is wrong, the ATM shows "Invalid PIN" instead of crashing → `except`

**3.** If the PIN is correct, you can continue with the transaction → `else`

**4.** After the transaction, the ATM returns your card and finishes the process → `finally`



To handle errors effectively in Python, we use different exception handling techniques such as `try`, `except`, `else`, `finally`, `raise`, `assert`, and custom exceptions.


# try

The `try` block contains code that may cause an error.

Python first executes the code inside the `try` block.

If an error occurs, Python can handle it using `except`.

### Real-Life Example

Trying to open a door:

- We try to open the door.
- If something goes wrong, we handle the problem.

```python
try:
    number = 10
    print(number)
except:
    print("An error occurred")
```

```python
try:
    number = Mani
    print(number)
except:
    print("An error occurred")
```

### explanation

- In example 1,
  
  `try` starts the block of code we want to test.

  `number = 10` stores 10.

  `print(number)` prints the value.

  since there is no error,the `except` block is not executed.

- In example 2,

  `try` starts the block of code we want to test.

  `number = mani` doesn't store mani.

  since there is an erroe, the `except` block is executed 

  

# except

The `except` block handles an error that occurs inside the `try` block.

### Real-Life Example

If we enter the wrong password, the system shows an error message instead of crashing.

```python
try:
    number = 7 / 0
except:
    print("Cannot divide by zero")
```

### Explanation

- 7 / 0 causes an error.
- Python stops executing the `try` block.
- The `except` block handles the error.
- Instead of stopping the program, it prints a message.

# else

The `else` block runs only when there is no error in the `try` block.

### Real-Life Example

If a payment is successful, we show "Payment Successful".

```python
try:
    number = 10 / 2
except:
    print("Something went wrong")
else:
    print("Division successful")
```

### Explanation

- 10 / 2 works successfully.
- No error occurs.
- Therefore, the `else` block executes.
- The `except` block is skipped

# finally

The `finally` block always executes, whether an error occurs or not.

### Real-Life Example

After using a file, database, or network connection, we may need to close it whether the operation succeeds or fails.

```python
try:
    number = 10 / 2
    print(number)
except:
    print("Error occurred")
finally:
    print("Program finished")
```

### Explanation

- The division is successful.
- 5.0 is printed.
- `finally` executes after the `try` block.
- `finally` executes regardless of whether an exception occurs.

# raise

The `raise` keyword is used to manually create an exception.

### Real-Life Example

During a transaction if a bank account has insufficient balance, the program can manually raise an error.

```python
age = 15

if age < 18:
    raise Exception("Age must be 18 or above")
```

### Explanation

- age is set to 15.
- The condition age < 18 is true.
- `raise` manually creates an exception.
- The message explains the problem.

# assert

The `assert` statement checks whether a condition is true.

If the condition is false, Python raises an `AssertionError`.

### Real-Life Example

We can check whether a student's marks are within the expected range.

```python
marks = 80

assert marks >= 0

print("Valid marks")
```

```python
# example with error

marks = 50

assert marks >= 80

print("Valid marks")
```

# Custom Exceptions

A custom exception is an exception created by the programmer for a specific situation.

We create a custom exception by creating a class that inherits from Python's built-in `Exception` class.

### Real-Life Example

A banking application can create a custom exception called `InsufficientBalanceError`.

```python
class InsufficientBalanceError(Exception):
    pass


balance = 500

if balance < 1000:
    raise InsufficientBalanceError("Insufficient balance")
```

### Explanation

- `InsufficientBalanceError` is our own custom exception class.
- It inherits from the built-in `Exception` class.
- `pass` means we do not need to add any extra code to the class.
- `balance` is set to `500`.
- The condition checks whether the balance is less than `1000`.
- Since the condition is true, we use `raise` to trigger our custom exception.
- The custom message `"Insufficient balance"` explains the problem.

# ValueError

`ValueError` occurs when a function receives a value of the correct type but the value itself is invalid.

### Real-World Example

If a user enters `"hello"` where the application expects a number, the value is invalid for that operation.

```python
number = int("hello")
```

### Explanation

- `int()` is used to convert a value into an integer.
- `"hello"` is a string, so its data type is valid for the `int()` function.
- `"hello"` does not contain a valid integer value.
-  Python raises `ValueError`.

# TypeError

`TypeError` occurs when an operation or function is used with an inappropriate data type.

### Real-World Example

Trying to add a number and a word directly is not a valid operation.

```python
number = 10
text = "Hello"

result = number + text
```

### Explanation

- `number` is an integer.
- `text` is a string.
- Python cannot directly add an integer and a string.
- Therefore, Python raises `TypeError`.

# IndexError

`IndexError` occurs when we try to access an index that does not exist in a sequence such as a list.

### Real-World Example

Imagine a row of only three seats. Trying to access seat number 6 would be invalid.

```python
names = ["Manikanta", "Rahul", "Ravi"]

print(names[5])
```

### Explanation

The list has three elements:

- Index `0` → Manikanta
- Index `1` → Rahul
- Index `2` → Ravi

There is no index `5`.

Python raises `IndexError`.

# KeyError

`KeyError` occurs when we try to access a dictionary using a key that does not exist.

### Real-World Example

If a student record contains `name` and `age`, but we try to access `course`, that key does not exist.

```python
student = {
    "name": "Manikanta",
    "age": 21
}

print(student["course"])
```

### Explanation

The dictionary contains the keys `name` and `age`.

The key `course` does not exist.

Therefore, Python raises `KeyError`.

# AttributeError

`AttributeError` occurs when an object does not have the attribute or method that we try to access.

### Real-World Example

A student object may have a `name` attribute, but if we try to access an attribute called `salary` that does not exist, Python raises an error.

```python
name = "Manikanta"

name.append("Reddy")
```

### Explanation

- `name` is a string.
- Strings do not have an `append()` method.
- `append()` is commonly used with lists.
-  Python raises `AttributeError`.

# ZeroDivisionError

# ImportError

- Python raises `ImportError`.
