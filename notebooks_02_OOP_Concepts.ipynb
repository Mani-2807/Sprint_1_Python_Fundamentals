# Object-Oriented Programming

**OOP (Object-Oriented Programming) is a way of writing computer code by organizing it around real-world "objects" rather than just a long list of logic and instructions.**
Instead of treating code like a step-by-step recipe, OOP treats a program like a world filled with distinct items that can interact with each other.

# 1. Class

## Understand the Concept

A class is a blueprint or template used to create objects.

A class defines:
- What data an object can have
- What actions an object can perform

A class itself is not a real object. It is a design/template from which objects are created.

## Real-Life Example

A house blueprint is a good example of a class.

The blueprint describes:
- Number of rooms
- Doors
- Windows
- Kitchen

The actual houses built from that blueprint are objects.

```python
class Student:
    name = "Manikanta"
    age = 21

print(Student.name)
print(Student.age)
```

## Code Explanation

- `class Student:` creates a class named Student.
- `name` is a data attribute of the class.
- `age` is another data attribute.
- `Student.name` accesses the name attribute.
- `Student.age` accesses the age attribute.

The class provides the structure that can later be used to create objects.

# 2. Object

## Understand the Concept

An object is an instance of a class.

When we create an object from a class, the object gets access to the attributes and methods defined by that class.

## Real-Life Example

A house blueprint is a class.

The actual houses built using that blueprint are objects.

syntax to create an object `object_name = classname()`

For example:

Blueprint → Class
Actual House → Object


```python
class Student:
    name = "Manikanta"
    age = 21


student1 = Student()

print(student1.name)
print(student1.age)
```

## Code Explanation

- `Student` is the class.
- `student1 = Student()` creates an object.
- `student1` is an object of the `Student` class.
- `student1.name` accesses the name through the object.
- `student1.age` accesses the age through the object.

#  3. `__init__()`

## Understand the Concept

`__init__()` is a special method that is automatically called when an object is created.

It is mainly used to initialize the object's attributes.

It is commonly called the constructor in Python, although technically `__new__()` is responsible for creating the object and `__init__()` initializes it.

## Real-Life Example

When a student joins a college, information such as:

- Name
- Age
- Roll number

can be initialized when the student object is created.

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age


student1 = Student("Manikanta", 21)

print(student1.name)
print(student1.age)
```

## Code Explanation

- `__init__()` runs automatically when `Student()` is called.
- `self` refers to the current object.
- `name` and `age` are parameters.
- `self.name` stores the name inside the object.
- `self.age` stores the age inside the object.
- `"Manikanta"` and `21` are passed when the object is created.

# 4. Instance Variables

## Understand the Concept

Instance variables are variables that belong to a particular object.

They are usually created using `self`.

Each object can have different values for its instance variables.

## Real-Life Example

Consider two students:

Student 1:
- Name = Manikanta
- Age = 21

Student 2:
- Name = Rahul
- Age = 22

Their names and ages are different, so these are instance variables.

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age


student1 = Student("Manikanta", 21)
student2 = Student("Rahul", 22)

print(student1.name, student1.age)
print(student2.name, student2.age)
```

## Code Explanation

- `self.name` is an instance variable.
- `self.age` is an instance variable.
- `student1` has its own name and age.
- `student2` has its own name and age.
- Changing one object's instance variable does not automatically change another object's variable.

# 5. Class Variables

## Understand the Concept

A class variable is a variable that belongs to the class rather than to one specific object.

It is shared by objects of that class unless an object creates its own variable with the same name.

## Real-Life Example

Suppose every student belongs to the same college.

The college name can be stored as a class variable because it is common to all students.

```python
class Student:

    college = "University of Hyderabad"

    def __init__(self, name):
        self.name = name


student1 = Student("Manikanta")
student2 = Student("Rahul")

print(student1.name)
print(student1.college)

print(student2.name)
print(student2.college)
```

## Code Explanation

- `college` is defined directly inside the class.
- Therefore, it is a class variable.
- Both objects can access the same class variable.
- `name` is an instance variable because it is created using `self`.

# 6. Instance Methods

## Understand the Concept

An instance method is a method that works with the data of a particular object.

It normally takes `self` as its first parameter.

## Real-Life Example

A student can perform an action such as introducing themselves.

The introduction depends on the particular student's data.

```python
class Student:

    def __init__(self, name):
        self.name = name

    def introduce(self):
        print("My name is", self.name)


student1 = Student("Manikanta")

student1.introduce()
```

## Code Explanation

- `introduce()` is an instance method.
- It receives `self`.
- `self.name` accesses the current object's name.
- `student1.introduce()` calls the method for `student1`.

# 7. Class Methods

## Understand the Concept

A class method is a method that works with the class itself rather than a particular object.

It is created using the `@classmethod` decorator.

Its first parameter is normally `cls`.

## Real-Life Example

Suppose a college changes its common college name.

Since the college name belongs to the class, a class method can be used to modify it.

```python
class Student:

    college = "University of Hyderabad"

    @classmethod
    def change_college(cls, new_college):
        cls.college = new_college


print(Student.college)

Student.change_college("IIT Hyderabad")

print(Student.college)
```

## Code Explanation

- `@classmethod` converts the method into a class method.
- `cls` refers to the class.
- `cls.college` accesses the class variable.
- `Student.change_college()` calls the class method.
- The class variable is updated.

# 8. Static Methods

## Understand the Concept

A static method is a method that does not need access to either the object or the class.

It is created using the `@staticmethod` decorator.

It behaves like a normal function but is logically grouped inside a class.

## Real-Life Example

A calculator operation such as addition does not need student or class information.

Therefore, it can be a static method.

```python
class Calculator:

    @staticmethod
    def add(a, b):
        return a + b


result = Calculator.add(10, 20)

print(result)
```

## Code Explanation

- `@staticmethod` creates a static method.
- The method does not use `self`.
- The method does not use `cls`.
- It simply receives two numbers and returns their sum.

# 9. @property

## Understand the Concept

`@property` allows a method to be accessed like an attribute.

It is useful when we want controlled access to data while keeping the syntax simple.

## Real-Life Example

Suppose a bank account has a balance.

Instead of directly exposing internal calculations, we can provide the balance through a property.

```python
class BankAccount:

    def __init__(self, balance):
        self._balance = balance

    @property
    def balance(self):
        return self._balance


account = BankAccount(5000)

print(account.balance)
```

## Code Explanation

- `_balance` stores the internal balance.
- `@property` makes `balance()` behave like an attribute.
- We access it using `account.balance`.
- We do not need to write `account.balance()`.

# 10. Encapsulation

## Understand the Concept

Encapsulation means combining data and methods inside a class and controlling how the data is accessed.

Python uses naming conventions and name mangling to indicate different levels of access.

Common conventions:

- `name` → public
- `_name` → protected by convention
- `__name` → private-like through name mangling

## Real-Life Example

A bank account should not allow anyone to directly modify sensitive account information.

Instead, methods can control how the data is accessed or changed.

```python
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance


account = BankAccount(5000)

print(account.get_balance())
```

## Code Explanation

- `__balance` uses double underscore name mangling.
- It is not intended to be accessed directly from outside the class.
- `get_balance()` provides controlled access.
- The data and method are kept together inside the class.

Python does not provide strict private access in the same way as some languages.
Double underscore mainly triggers name mangling.

# 11. Abstraction

## Understand the Concept

Abstraction means hiding implementation details and showing only the essential functionality.

Python provides abstract classes using the `abc` module.

## Real-Life Example

When we drive a car, we use:

- Steering
- Brake
- Accelerator

We do not need to know the internal engine implementation to drive the car.

```python
from abc import ABC, abstractmethod


class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass


class Dog(Animal):

    def sound(self):
        print("Dog barks")


dog = Dog()

dog.sound()
```

## Explanation

- `ABC` creates an abstract base class.
- `@abstractmethod` defines a method that must be implemented.
- `Animal` defines `sound()`.
- `Dog` provides the implementation.

# 12. Inheritance

## Understand the Concept

Inheritance allows a child class to use functionality from a parent class.

## Real-Life Example

A car is a type of vehicle.

Vehicle → Parent

Car → Child

```python
class Vehicle:

    def start(self):
        print("Vehicle starts")


class Car(Vehicle):
    pass


car = Car()

car.start()
```

## Explanation

- `Vehicle` is the parent class.
- `Car` is the child class.
- `Car(Vehicle)` creates inheritance.
- The `Car` object can use `start()` from `Vehicle`.

# 13. Multiple Inheritance

## Understand the Concept

Multiple inheritance occurs when one child class inherits from more than one parent class.

## Real-Life Example

A person can have both programming and speaking skills.

```python
class Programmer:

    def code(self):
        print("Writing code")


class Speaker:

    def speak(self):
        print("Speaking")


class Person(Programmer, Speaker):
    pass


person = Person()

person.code()
person.speak()
```

## Explanation

- `Programmer` is one parent.
- `Speaker` is another parent.
- `Person` inherits from both.
- Therefore, `Person` can use methods from both classes.

# 14. Multilevel Inheritance

## Understand the Concept

Multilevel inheritance means inheritance happens through multiple levels.

Example:

Grandparent → Parent → Child

```python
class Grandparent:

    def house(self):
        print("Grandparent has a house")


class Parent(Grandparent):
    pass


class Child(Parent):
    pass


child = Child()

child.house()
```

## Explanation

- `Parent` inherits from `Grandparent`.
- `Child` inherits from `Parent`.
- Therefore, `Child` can access the method of `Grandparent`.

# 15. Hierarchical Inheritance

## Understand the Concept

Hierarchical inheritance occurs when multiple child classes inherit from the same parent class.

Example:

        Vehicle
        /     \
      Car     Bike

```python
class Vehicle:

    def start(self):
        print("Vehicle starts")


class Car(Vehicle):
    pass


class Bike(Vehicle):
    pass


car = Car()
bike = Bike()

car.start()
bike.start()
```

## Explanation

- `Vehicle` is the parent class.
- `Car` is one child.
- `Bike` is another child.
- Both inherit `start()` from `Vehicle`.

# 16. Polymorphism

## Understand the Concept

Polymorphism means "many forms".

The same method name can behave differently for different objects.

## Real-Life Example

Different animals make different sounds.

Dog → Bark

Cat → Meow

```python
class Dog:

    def sound(self):
        print("Bark")


class Cat:

    def sound(self):
        print("Meow")


dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
```

## Explanation

Both classes have the same method name:

sound()

But each class gives a different result.

This is polymorphism.

# 17. Method Overriding

## Understand the Concept

Method overriding occurs when a child class provides its own version of a method already present in the parent class.

## Real-Life Example

A parent class provides a general greeting.

A child class changes the greeting.

```python
class Parent:

    def message(self):
        print("Hello from Parent")


class Child(Parent):

    def message(self):
        print("Hello from Child")


child = Child()

child.message()
```

## Explanation

- `Parent` has a `message()` method.
- `Child` also defines `message()`.
- The child's method overrides the parent's method.

# 18. Method Overloading

## Understand the Concept

Method overloading means using the same method name with different numbers or types of parameters.

Python does not support traditional method overloading like Java or C++.

We can achieve similar behavior using default arguments.

## Real-Life Example

An addition function can work with two or three numbers.

```python
class Calculator:

    def add(self, a, b, c=0):
        return a + b + c


calculator = Calculator()

print(calculator.add(10, 20))
print(calculator.add(10, 20, 30))
```

## Explanation

- `c` has a default value of `0`.
- If we provide two numbers, `c` remains 0.
- If we provide three numbers, all three are added.
- This provides overloading-like behavior.

# 19. super()

## Understand the Concept

`super()` is used to access methods from the parent class.

It is commonly used to call the parent's `__init__()` method.

## Real-Life Example

A child class can use the parent's initialization and then add its own data.

```python
class Person:

    def __init__(self, name):
        self.name = name


class Student(Person):

    def __init__(self, name):
        super().__init__(name)


student = Student("Manikanta")

print(student.name)
```

## Explanation

- `Person` is the parent.
- `Student` is the child.
- `super().__init__(name)` calls the parent's `__init__()`.
- The parent initializes `self.name`.

# 20. Dunder Methods

Examples:

__init__()
__str__()
__repr__()
__len__()

Python automatically uses these methods for certain operations.

```python
class Student:

    def __init__(self, name):
        self.name = name


student = Student("Manikanta")

print(student.name)
```

## Explanation

`__init__()` is a dunder method.

Dunder methods allow us to define how objects behave with Python's built-in operations.

# 21. `__str__()`

## Understand the Concept

`__str__()` defines a human-readable representation of an object.

It is commonly used when printing an object.

```python
class Student:

    def __init__(self, name):
        self.name = name

    def __str__(self):
        return self.name


student = Student("Manikanta")

print(student)
```

## Explanation

- `__str__()` returns a string.
- `print(student)` automatically uses `__str__()`.
- It gives a simple human-readable representation.

# 22. `__repr__()`

## Understand the Concept

`__repr__()` provides a developer-oriented representation of an object.

It is useful when inspecting an object.

```python
class Student:

    def __init__(self, name):
        self.name = name

    def __repr__(self):
        return f"Student('{self.name}')"


student = Student("Manikanta")

print(repr(student))
```

## Explanation

- `__repr__()` returns a string representation.
- `repr(student)` calls `__repr__()`.
- The result gives useful information about the object.

# 23. `__len__()`

## Understand the Concept

`__len__()` defines what happens when we use `len()` on an object.

## Real-Life Example

A shopping cart can use `__len__()` to tell us how many items it contains.

```python
class ShoppingCart:

    def __init__(self):
        self.items = ["Pen", "Book", "Bag"]

    def __len__(self):
        return len(self.items)


cart = ShoppingCart()

print(len(cart))
```

## Explanation

- `items` contains three items.
- `__len__()` returns the number of items.
- `len(cart)` calls `__len__()`.

# 24. Composition

## Understand the Concept

Composition represents a strong "has-a" relationship.

One class contains an object of another class.

## Real-Life Example

A car has an engine.

```python
class Engine:

    def start(self):
        print("Engine starts")


class Car:

    def __init__(self):
        self.engine = Engine()


car = Car()

car.engine.start()
```

## Explanation

- `Engine` is one class.
- `Car` creates an `Engine` object inside itself.
- Therefore, Car has an Engine.
- This is composition.

# 25. Aggregation

## Understand the Concept

Aggregation is a weaker "has-a" relationship.

The object being used can exist independently.

## Real-Life Example

A teacher can exist without a department.

A department can have a teacher.

```python
class Teacher:

    def __init__(self, name):
        self.name = name


class Department:

    def __init__(self, teacher):
        self.teacher = teacher


teacher = Teacher("Ravi")

department = Department(teacher)

print(department.teacher.name)
```

## Explanation

- The teacher object is created first.
- The department receives that teacher object.
- The teacher can exist independently.
- Therefore, this is aggregation.

# 26. Association

## Understand the Concept

Association means two independent objects are connected or interact with each other.

## Real-Life Example

A teacher teaches a student.

Both teacher and student can exist independently.

```python
class Teacher:

    def teach(self):
        print("Teacher is teaching")


class Student:

    def learn(self):
        print("Student is learning")


teacher = Teacher()
student = Student()

teacher.teach()
student.learn()
```

## Explanation

- Teacher and Student are separate classes.
- Their objects can exist independently.
- They represent related entities.
- This is an example of association.

# 27. Dataclasses

## Understand the Concept

A dataclass is useful when a class mainly stores data.

Python can automatically create methods such as `__init__()` and `__repr__()`.

## Real-Life Example

We can use a dataclass to store student information.

```python
from dataclasses import dataclass


@dataclass
class Student:
    name: str
    age: int


student = Student("Manikanta", 21)

print(student)
```

## Explanation

- `@dataclass` makes the class a dataclass.
- `name` and `age` are data fields.
- Python automatically creates an `__init__()` method.
- Python also provides a useful representation of the object.

# 28. Singleton

## Understand the Concept

Singleton is a design pattern that is intended to allow only one instance of a class.

It is not a Python keyword.

## Simple Example

We can use a class variable to keep track of the number of objects.

```python
class Singleton:

    instance = None

    def __new__(cls):
        if cls.instance is None:
            cls.instance = super().__new__(cls)

        return cls.instance


object1 = Singleton()
object2 = Singleton()

print(object1 is object2)
```

## Explanation

- `instance` stores the object.
- `__new__()` controls object creation.
- The first time, an object is created.
- The second time, the same object is returned.
- `object1 is object2` returns `True`.
