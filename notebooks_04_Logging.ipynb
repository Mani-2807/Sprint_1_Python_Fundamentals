# Logging

# What is Logging?

Logging is the process of recording information about what is happening inside a program while it is running.

Python provides a built-in `logging` module for creating and managing log messages.

Logs can help us understand:

- What the program is doing
- Whether an operation was successful
- When a warning occurs
- When an error occurs
- What happened before a problem occurred

### Real-World Example

Think about a security camera in a building.

The camera records what happens over time.

Similarly, logging records important events that happen inside a program.


```python
import logging

logging.basicConfig(level=logging.INFO)

logging.info("Program started")
```

### Explanation

- `import logging` imports Python's built-in logging module.
- `logging.basicConfig()` sets the basic logging configuration.
- `level=logging.INFO` tells Python to display INFO-level messages and above.
- `logging.info()` creates an INFO log message.
- `"Program started"` is the message recorded in the log.


# Why Use Logging?

Logging is used to monitor and understand the behavior of a program.

It helps developers:

- Track program execution
- Find errors
- Understand warnings
- Debug problems
- Monitor applications
- Maintain a history of important events

### Real-World Example

Imagine an online shopping application.

When a customer places an order, the application can record:

- Order started
- Payment successful
- Order created
- Delivery started
- Error occurred

These records help developers understand what happened if something goes wrong.

```python
import logging

logging.basicConfig(level=logging.INFO)

logging.info("Order started")
logging.info("Payment successful")
logging.info("Order created")
```

### Explanation

Each `logging.info()` records an important event.

Instead of checking the entire program manually, developers can look at the logs to understand what happened.



# print() vs Logging

Both `print()` and logging can display information, but they are used differently.

`print()` is mainly used for simple output while developing or testing a program.

Logging is designed for monitoring and recording application events.

### Real-World Example

`print()` is like telling someone what is happening right now.

Logging is like maintaining an official record of what happened.

### Main Difference

- `print()` → simple display
- Logging → structured application records

```python
import logging

logging.basicConfig(level=logging.INFO)

print("User logged in")

logging.info("User logged in")
```

### Explanation

- `print()` directly displays text.
- `logging.info()` creates a log message with a logging level.
- Logging becomes more useful when applications become larger.
- Logging can also write messages to files, include timestamps, and separate messages


# Logging Levels

Logging levels tell us how important a log message is.

Python provides five commonly used logging levels:

1. `DEBUG`
2. `INFO`
3. `WARNING`
4. `ERROR`
5. `CRITICAL`

The levels represent increasing importance.

DEBUG
↓
INFO
↓
WARNING
↓
ERROR
↓
CRITICAL

### Real-World Example

Think about traffic signals:

- DEBUG → small information
- INFO → normal situation
- WARNING → something needs attention
- ERROR → something went wrong
- CRITICAL → very serious problem

```python
import logging

logging.basicConfig(level=logging.DEBUG)

logging.debug("Debug message")
logging.info("Info message")
logging.warning("Warning message")
logging.error("Error message")
logging.critical("Critical message")
```

### Explanation

- `DEBUG` provides detailed information.
- `INFO` provides normal information.
- `WARNING` indicates a possible problem.
- `ERROR` indicates that an operation failed.
- `CRITICAL` indicates a very serious problem.


# DEBUG

`DEBUG` is the lowest commonly used logging level.

It is used for detailed information that helps developers understand and troubleshoot a program.

### Real-World Example

A developer may want to know which function was called or what value a variable contains while debugging a program.

```python
import logging

logging.basicConfig(level=logging.DEBUG)

logging.debug("Checking user information")
```

### Explanation

- `logging.DEBUG` sets the logging level to DEBUG.
- `logging.debug()` creates a DEBUG message.
- DEBUG messages are mainly useful during development and troubleshooting.



# INFO

`INFO` is used to record normal events showing that the program is working as expected.

### Real-World Example

An application can record that a user successfully logged in.

```python
import logging

logging.basicConfig(level=logging.INFO)

logging.info("User logged in successfully")
```

### Explanation

`logging.info()` records a normal event in the application.


# WARNING

`WARNING` indicates that something unexpected happened or may cause a problem in the future.

The program can usually continue running.

### Real-World Example

A website may warn that a user's storage is almost full.

```python
import logging

logging.basicConfig(level=logging.WARNING)

logging.warning("Storage is almost full")
```

### Explanation

- `logging.warning()` creates a WARNING message.
- A warning does not necessarily mean the program has stopped.
- It tells developers or users that something needs attention.



# ERROR

`ERROR` indicates that an operation failed or a problem occurred in the program.

### Real-World Example

If a payment fails, the application can record an ERROR message.

```python
import logging

logging.basicConfig(level=logging.ERROR)

logging.error("Payment failed")
```

### Explanation

- `logging.error()` records an ERROR message.
- It tells us that something went wrong during program execution.


# CRITICAL

`CRITICAL` represents a very serious problem that may prevent the application from continuing normally.

### Real-World Example

If an important database server becomes unavailable, the application may record a CRITICAL message.

```python
import logging

logging.basicConfig(level=logging.CRITICAL)

logging.critical("Database server is unavailable")
```

### Explanation

`logging.critical()` records a very serious problem.

CRITICAL messages are more severe than ERROR messages.


# Formatters

A formatter controls how log messages are displayed.

We can include useful information such as:

- Time
- Logging level
- Message

### Real-World Example

Instead of writing only:

Payment successful

we can record:

2026-09-24 10:30:00 - INFO - Payment successful

This gives us more information about the event.

```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

logging.info("Payment successful")
```

### Explanation

- `level=logging.INFO` sets the minimum logging level.
- `format=` defines how the log message should look.
- `%(asctime)s` displays the time.
- `%(levelname)s` displays the logging level.
- `%(message)s` displays the actual log message.


# File Handlers

A file handler is used to save log messages into a file instead of displaying them only on the screen.

### Real-World Example

A company's application may save error logs in `application.log`.

Developers can check this file later to investigate problems.

```python
import logging

logging.basicConfig(
    filename="application.log",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

logging.info("Application started")
logging.error("An error occurred")
```

### Explanation

- `filename="application.log"` tells Python to store logs in this file.
- `level=logging.INFO` records INFO and higher-level messages.
- The formatter controls the appearance of the log messages.
- `logging.info()` records a normal event.
- `logging.error()` records an error.


# Console Handlers

A console handler sends log messages to the console or terminal.

This is useful when we want to see logs while the program is running.

### Real-World Example

A developer running an application can see important events directly in the terminal.

```python
import logging

logger = logging.getLogger("my_logger")
logger.setLevel(logging.INFO)

console_handler = logging.StreamHandler()

logger.addHandler(console_handler)

logger.info("Application started")
```

### Explanation

- `getLogger()` creates a logger.
- `setLevel()` sets the logging level.
- `StreamHandler()` creates a handler for console output.
- `addHandler()` connects the handler to the logger.
- `logger.info()` creates the log message.


# Rotating Logs

Rotating logs means creating new log files when the current log file reaches a certain size or time limit.

This prevents one log file from becoming extremely large.

### Real-World Example

A busy website may generate thousands of log messages every day.

Instead of storing everything in one huge file, logs can be divided into multiple files.

```python
import logging
from logging.handlers import RotatingFileHandler

handler = RotatingFileHandler(
    "app.log",
    maxBytes=1000,
    backupCount=3
)

logger = logging.getLogger("app")
logger.setLevel(logging.INFO)

logger.addHandler(handler)

logger.info("Application started")
```

### Explanation

- `RotatingFileHandler` manages log file rotation.
- `"app.log"` is the main log file.
- `maxBytes=1000` means rotation can happen when the file reaches approximately 1000 bytes.
- `backupCount=3` keeps up to three backup log files.
- This helps prevent log files from becoming too large.


# Logging Best Practices

Good logging makes applications easier to understand, debug, and maintain.

### Important Practices

1. Use appropriate logging levels.
2. Use meaningful log messages.
3. Do not log sensitive information such as passwords.
4. Use formatters to include useful information.
5. Store important logs in files when required.
6. Avoid excessive logging.
7. Use DEBUG logs mainly for development and troubleshooting.

### Real-World Example

In a banking application, we can log:

- Login successful
- Transaction completed
- Payment failed

But we should never log sensitive information such as a customer's password.
